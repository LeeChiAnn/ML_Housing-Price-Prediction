# 房价预测训练选择器

这个 Notebook 允许你在 AI Studio 中直接选择运行两种版本中的任意一种：

- NumPy 版本：`work/1-2-build_neural_network_using_numpy.py`
- Paddle 版本：`work/1-3-build_neural_network_using_paddle.py`

你只需要修改下面的 `BACKEND` 变量，然后运行后续单元格即可。

```python
BACKEND = 'paddle'  # 可选：'numpy' 或 'paddle'
print(f'当前选择的版本：{BACKEND}')
```

```python
import importlib.util
from pathlib import Path

script_map = {
    'numpy': Path('./work/1-2-build_neural_network_using_numpy.py'),
    'paddle': Path('./work/1-3-build_neural_network_using_paddle.py'),
}

selected_script = script_map[BACKEND]
print(f'将加载脚本：{selected_script}')

spec = importlib.util.spec_from_file_location(f'{BACKEND}_trainer', selected_script)
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)

print('加载成功。')
print('可调用函数：', [name for name in ['train', 'validation', 'load_data'] if hasattr(module, name)])
```

## 修改训练参数

下面的参数会直接影响训练效果。

- `EPOCH_NUM`：训练轮数
- `BATCH_SIZE`：每批样本数量
- `LEARNING_RATE`：学习率

如果你想做实验，可以在下面直接修改。

```python
if hasattr(module, 'EPOCH_NUM'):
    module.EPOCH_NUM = 10
if hasattr(module, 'BATCH_SIZE'):
    module.BATCH_SIZE = 10
if hasattr(module, 'LEARNING_RATE'):
    module.LEARNING_RATE = 0.01

print('当前训练参数：')
for key in ['EPOCH_NUM', 'BATCH_SIZE', 'LEARNING_RATE']:
    if hasattr(module, key):
        print(f'{key} = {getattr(module, key)}')
```

```python
# 运行当前选择的版本
if hasattr(module, 'train'):
    module.train()
else:
    print('当前版本没有 train() 函数。')
```

```python
# 可选：查看预测效果
if hasattr(module, 'validation'):
    module.validation()
else:
    print('当前版本没有 validation() 函数。')
```
